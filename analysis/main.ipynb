```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
df = pd.read_csv("output/results.csv")
```

```python
col_especie = "especie"
col_atributo = "numero"
```

```python
df
```

```python
print("--- Estatísticas Descritivas por Espécie ---")
desc = df.groupby(col_especie)[col_atributo].agg(['count', 'mean', 'std', 'min', 'max'])
desc.to_csv("estatisticas_descritivas.csv")
```

```python
# 4. Visualização Gráfica
fig, axes = plt.subplots(1, figsize=(8, 5))

# Boxplot
sns.boxplot(data=df, x=col_especie, y=col_atributo, ax=axes, palette="Set2")
axes.set_title("Distribuição do Nº de Pixels de Borda por Espécie")
axes.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 4. Visualização Gráfica
fig, axes = plt.subplots(1, figsize=(8, 5))

# Curvas de Densidade (KDE)
sns.kdeplot(data=df, x=col_atributo, hue=col_especie, common_norm=False, ax=axes, fill=True, alpha=0.3)
axes.set_title("Sobreposição das Densidades de Probabilidade")

plt.tight_layout()
plt.show()
```
